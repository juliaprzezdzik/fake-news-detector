```python
# =====================================================================
# GPT zero-shot + few-shot na LIAR (2 klasy i 6 klas) - jedna komorka Colab
# Robi 4 eksperymenty: (2kl,zero) (2kl,few) (6kl,zero) (6kl,few)
# Zapisuje metryki, confusion matrix PNG, predykcje CSV i tabele porownawcza
# =====================================================================

# --------- 0. Instalacja + import ----------
import subprocess, sys
subprocess.run([sys.executable, "-m", "pip", "-q", "install",
              "openai>=1.30.0", "scikit-learn", "pandas", "tqdm", "matplotlib"], check=True)

import os, re, json, hashlib, sqlite3, asyncio, zipfile, urllib.request, getpass, random, time
from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tqdm.auto import tqdm
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.metrics import (precision_recall_fscore_support, classification_report,
                            confusion_matrix, accuracy_score)
from openai import AsyncOpenAI

# --------- 1. KONFIG ----------
MODEL              = "gpt-4o-mini"   # albo "gpt-4o", "gpt-4.1-mini"
N_TEST             = 0               # 0 = caly test set, inaczej probkuje N
K_PER_CLASS        = 1               # ile przykladow few-shot na klase
FEWSHOT_STRATEGY   = "knn"           # "knn" (TF-IDF cosine) lub "random"
SELF_CONSISTENCY_N = 1               # 1 = bez, >=3 = glosowanie wiekszosciowe
CONCURRENCY        = 20              # rownolegle zapytania API
SEED               = 42
OUT_DIR            = "results"
CACHE_DB           = "gpt_cache.sqlite"

EXPERIMENTS = [
  {"num_classes": 2, "mode": "zero-shot"},
  {"num_classes": 2, "mode": "few-shot"},
  {"num_classes": 6, "mode": "zero-shot"},
  {"num_classes": 6, "mode": "few-shot"},
]

# --------- 2. Klucz API ----------
if not os.environ.get("OPENAI_API_KEY"):
  try:
      from google.colab import userdata
      os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
  except Exception:
      os.environ["OPENAI_API_KEY"] = getpass.getpass("OPENAI_API_KEY: ")

random.seed(SEED); np.random.seed(SEED)
os.makedirs(OUT_DIR, exist_ok=True)

# --------- 3. Pobranie LIAR ----------
DATA_DIR = "data"
ZIP_PATH = os.path.join(DATA_DIR, "liar_dataset.zip")
os.makedirs(DATA_DIR, exist_ok=True)
if not os.path.exists(os.path.join(DATA_DIR, "train.tsv")):
  urllib.request.urlretrieve("https://www.cs.ucsb.edu/~william/data/liar_dataset.zip", ZIP_PATH)
  with zipfile.ZipFile(ZIP_PATH) as z:
      z.extractall(DATA_DIR)

COLUMNS = ["id","label","statement","subject","speaker","job_title","state","party",
          "barely_true_counts","false_counts","half_true_counts","mostly_true_counts",
          "pants_on_fire_counts","context"]

def load_split(name):
  df = pd.read_csv(os.path.join(DATA_DIR, f"{name}.tsv"), sep="\t", header=None, names=COLUMNS)
  return df.drop(columns=["id","job_title","state"])

train_raw = load_split("train")
test_raw  = load_split("test")
print(f"train: {train_raw.shape}, test: {test_raw.shape}")

# --------- 4. Combined text (identycznie jak w notebooku RoBERTa) ----------
def combine_text(row):
  speaker = row.get("speaker") if pd.notna(row.get("speaker")) else "unknown"
  party   = row.get("party")   if pd.notna(row.get("party"))   else "unknown"
  def _n(x):
      try:
          v = float(x)
          return 0 if pd.isna(v) else int(v)
      except (TypeError, ValueError):
          return 0
  lies   = _n(row.get("false_counts")) + _n(row.get("pants_on_fire_counts"))
  truths = _n(row.get("mostly_true_counts")) + _n(row.get("half_true_counts"))
  parts = [f"{speaker} ({party}, past lies: {lies}, past truths: {truths}): {row['statement']}"]
  if pd.notna(row.get("subject")):
      parts.append("Subject: " + str(row["subject"]))
  if pd.notna(row.get("context")):
      parts.append("Context: " + str(row["context"]))
  return " ".join(parts)

train_raw["combined"] = train_raw.apply(combine_text, axis=1)
test_raw["combined"]  = test_raw.apply(combine_text, axis=1)

# --------- 5. Etykiety dla 2kl i 6kl ----------
LABELS_6 = ["pants-fire", "false", "barely-true", "half-true", "mostly-true", "true"]
LABEL_DESC_6 = {
  "pants-fire":  "razaco falszywe, absurdalne klamstwo",
  "false":       "falszywe",
  "barely-true": "zawiera ziarno prawdy, ale w istotny sposob mylace",
  "half-true":   "czesciowo prawdziwe, pomija istotne fakty",
  "mostly-true": "w wiekszosci prawdziwe, drobne niescislosci",
  "true":        "prawdziwe",
}
BIN_MAP = {"pants-fire":"fake","false":"fake","barely-true":"fake",
          "half-true":"real","mostly-true":"real","true":"real"}
LABELS_2 = ["fake", "real"]

def prep_split(df, num_classes):
  df = df[df["label"].isin(LABELS_6)].copy()
  if num_classes == 2:
      df["y_str"] = df["label"].map(BIN_MAP)
      labels = LABELS_2
  else:
      df["y_str"] = df["label"]
      labels = LABELS_6
  df["y"] = df["y_str"].map({l:i for i,l in enumerate(labels)})
  return df.reset_index(drop=True), labels

# --------- 6. Prompty ----------
def build_system_prompt(labels, num_classes):
  if num_classes == 2:
      guide = ("- fake: wypowiedz falszywa lub w wiekszosci mylaca\n"
                "- real: wypowiedz prawdziwa lub w wiekszosci prawdziwa")
  else:
      guide = "\n".join(f"- {l}: {LABEL_DESC_6[l]}" for l in labels)
  return ("Jestes klasyfikatorem prawdziwosci wypowiedzi politycznych ze zbioru LIAR. "
          f"Przypisujesz wypowiedzi DOKLADNIE jedna z {len(labels)} etykiet:\n{guide}\n"
          f"Odpowiadasz WYLACZNIE jedna etykieta z listy: {', '.join(labels)}. "
          "Bez wyjasnien, bez interpunkcji.")

def fmt_user(combined, labels):
  return f"Wypowiedz:\n{combined}\n\nEtykieta ({', '.join(labels)}):"

def build_messages(combined, labels, num_classes, examples=None):
  msgs = [{"role":"system","content":build_system_prompt(labels, num_classes)}]
  if examples:
      for ex in examples:
          msgs.append({"role":"user","content":fmt_user(ex["combined"], labels)})
          msgs.append({"role":"assistant","content":ex["y_str"]})
  msgs.append({"role":"user","content":fmt_user(combined, labels)})
  return msgs

# --------- 7. Few-shot selektor ----------
class FewShotSelector:
  def __init__(self, train_df, labels, strategy="knn"):
      self.train_df = train_df
      self.labels = labels
      self.strategy = strategy
      if strategy == "knn":
          self.vec = TfidfVectorizer(ngram_range=(1,2), min_df=2, max_features=50000)
          self.train_mat = self.vec.fit_transform(train_df["combined"].tolist())
      self.by_label = {l: train_df[train_df.y_str == l].reset_index(drop=True) for l in labels}
      self.by_label_idx = {l: train_df.index[train_df.y_str == l].to_numpy() for l in labels}

  def select(self, query_combined, k_per_class, seed):
      out = []
      if self.strategy == "random":
          for l in self.labels:
              pool = self.by_label[l]
              if len(pool) == 0: continue
              out.extend(pool.sample(min(k_per_class, len(pool)), random_state=seed).to_dict("records"))
      else:  # knn per klasa
          q = self.vec.transform([query_combined])
          sims = cosine_similarity(q, self.train_mat).ravel()
          for l in self.labels:
              idxs = self.by_label_idx[l]
              if len(idxs) == 0: continue
              top = idxs[np.argsort(-sims[idxs])[:k_per_class]]
              out.extend(self.train_df.iloc[top].to_dict("records"))
      random.Random(seed).shuffle(out)
      return out

# --------- 8. Cache SQLite ----------
def cache_init():
  con = sqlite3.connect(CACHE_DB)
  con.execute("CREATE TABLE IF NOT EXISTS c (k TEXT PRIMARY KEY, v TEXT)")
  con.commit(); return con

def cache_key(model, messages, temperature, sample_idx):
  blob = json.dumps([model, messages, temperature, sample_idx], ensure_ascii=False)
  return hashlib.sha1(blob.encode()).hexdigest()

def cache_get(con, k):
  r = con.execute("SELECT v FROM c WHERE k=?", (k,)).fetchone()
  return r[0] if r else None

def cache_put(con, k, v):
  con.execute("INSERT OR REPLACE INTO c VALUES (?,?)", (k, v)); con.commit()

# --------- 9. Parsowanie odpowiedzi ----------
def parse_label(text, labels):
  t = text.strip().lower().strip(".,;:!?\"'` ")
  if t in labels: return t
  for l in sorted(labels, key=len, reverse=True):
      if l in t: return l
  return None

# --------- 10. Async runner ----------
async def call_one(client, sem, con, model, messages, temperature, sample_idx, max_retries=4):
  k = cache_key(model, messages, temperature, sample_idx)
  hit = cache_get(con, k)
  if hit is not None:
      return hit
  async with sem:
      for attempt in range(max_retries):
          try:
              resp = await client.chat.completions.create(
                  model=model, messages=messages,
                  max_tokens=8, temperature=temperature, seed=SEED+sample_idx)
              txt = resp.choices[0].message.content or ""
              cache_put(con, k, txt)
              return txt
          except Exception as e:
              if attempt == max_retries-1:
                  print(f"[API error] {e}")
                  return ""
              await asyncio.sleep(2 ** attempt + random.random())

async def classify_batch(client, con, model, all_messages, labels, sc_n):
  sem = asyncio.Semaphore(CONCURRENCY)
  if sc_n <= 1:
      tasks = [call_one(client, sem, con, model, m, 0.0, 0) for m in all_messages]
      outs = await asyncio.gather(*[t for t in tqdm_asyncio_wrap(tasks)])
      return [parse_label(o, labels) for o in outs]
  # self-consistency: sc_n probek z T=0.7
  all_tasks = []
  for i, m in enumerate(all_messages):
      for s in range(sc_n):
          all_tasks.append(call_one(client, sem, con, model, m, 0.7, s))
  raw = await asyncio.gather(*[t for t in tqdm_asyncio_wrap(all_tasks)])
  preds = []
  for i in range(len(all_messages)):
      votes = [parse_label(raw[i*sc_n + s], labels) for s in range(sc_n)]
      votes = [v for v in votes if v]
      preds.append(Counter(votes).most_common(1)[0][0] if votes else None)
  return preds

def tqdm_asyncio_wrap(tasks):
  # zwykly tqdm na coroutine list
  pbar = tqdm(total=len(tasks))
  async def wrap(t):
      r = await t; pbar.update(1); return r
  out = [wrap(t) for t in tasks]
  return out

# --------- 11. Ewaluacja + zapis ----------
def evaluate_and_save(y_true, y_pred_str, labels, tag, test_df):
  label2id = {l:i for i,l in enumerate(labels)}
  y_pred = [label2id.get(p, -1) for p in y_pred_str]
  mask = [p != -1 for p in y_pred]
  yt = np.array([y for y,m in zip(y_true, mask) if m])
  yp = np.array([y for y,m in zip(y_pred, mask) if m])
  n_unparsed = sum(1 for m in mask if not m)

  idx = list(range(len(labels)))
  per_p, per_r, per_f1, per_sup = precision_recall_fscore_support(
      yt, yp, labels=idx, average=None, zero_division=0)
  macro = precision_recall_fscore_support(yt, yp, labels=idx, average="macro", zero_division=0)
  weighted = precision_recall_fscore_support(yt, yp, labels=idx, average="weighted", zero_division=0)
  acc = accuracy_score(yt, yp)
  cm = confusion_matrix(yt, yp, labels=idx)

  report_txt = classification_report(yt, yp, labels=idx, target_names=labels, zero_division=0)
  print(f"\n=== {tag} ===  acc={acc:.3f}  macro-F1={macro[2]:.3f}  unparsed={n_unparsed}")
  print(report_txt)
  print("Confusion matrix (rows=true, cols=pred):")
  print(pd.DataFrame(cm, index=labels, columns=labels))

  # heatmapa
  fig, ax = plt.subplots(figsize=(1.2*len(labels)+2, 1.2*len(labels)+1))
  im = ax.imshow(cm, cmap="Blues")
  ax.set_xticks(range(len(labels))); ax.set_xticklabels(labels, rotation=45, ha="right")
  ax.set_yticks(range(len(labels))); ax.set_yticklabels(labels)
  for i in range(len(labels)):
      for j in range(len(labels)):
          ax.text(j, i, str(cm[i,j]), ha="center", va="center",
                  color="white" if cm[i,j] > cm.max()/2 else "black")
  ax.set_xlabel("Predicted"); ax.set_ylabel("True"); ax.set_title(tag)
  fig.colorbar(im); fig.tight_layout()
  png = os.path.join(OUT_DIR, f"cm_{tag}.png")
  fig.savefig(png, dpi=120); plt.show(); plt.close(fig)

  # predykcje CSV
  test_df = test_df.copy()
  test_df["pred"] = y_pred_str
  test_df["correct"] = test_df["y_str"] == test_df["pred"]
  csv_path = os.path.join(OUT_DIR, f"preds_{tag}.csv")
  test_df[["speaker","party","subject","statement","y_str","pred","correct"]].to_csv(csv_path, index=False)

  # analiza bledow: top mylonych subjectow
  err = test_df[~test_df["correct"]]
  top_err_subj = err["subject"].value_counts().head(10).to_dict()

  metrics = {
      "tag": tag, "n_total": len(y_true), "n_unparsed": int(n_unparsed),
      "accuracy": float(acc),
      "macro": {"precision":float(macro[0]),"recall":float(macro[1]),"f1":float(macro[2])},
      "weighted": {"precision":float(weighted[0]),"recall":float(weighted[1]),"f1":float(weighted[2])},
      "per_class": [
          {"label":l,"precision":float(per_p[i]),"recall":float(per_r[i]),
            "f1":float(per_f1[i]),"support":int(per_sup[i])}
          for i,l in enumerate(labels)
      ],
      "confusion_matrix": cm.tolist(),
      "labels": labels,
      "top_error_subjects": top_err_subj,
      "cm_png": png, "preds_csv": csv_path,
  }
  with open(os.path.join(OUT_DIR, f"metrics_{tag}.json"), "w") as f:
      json.dump(metrics, f, ensure_ascii=False, indent=2)
  return metrics

# --------- 12. Glowna petla po eksperymentach ----------
async def main():
  client = AsyncOpenAI()
  con = cache_init()
  summary_rows = []

  for exp in EXPERIMENTS:
      nc, mode = exp["num_classes"], exp["mode"]
      tag = f"{MODEL}_{nc}cl_{mode}_{FEWSHOT_STRATEGY if mode=='few-shot' else 'na'}_k{K_PER_CLASS}_sc{SELF_CONSISTENCY_N}"
      print(f"\n############ {tag} ############")

      train_df, labels = prep_split(train_raw, nc)
      test_df,  _      = prep_split(test_raw,  nc)
      if N_TEST > 0:
          test_df = test_df.sample(min(N_TEST, len(test_df)), random_state=SEED).reset_index(drop=True)
      print(f"Rozklad testu: {test_df['y_str'].value_counts().to_dict()}")

      selector = FewShotSelector(train_df, labels, FEWSHOT_STRATEGY) if mode == "few-shot" else None

      all_messages = []
      for _, row in test_df.iterrows():
          if mode == "zero-shot":
              msgs = build_messages(row["combined"], labels, nc)
          else:
              ex = selector.select(row["combined"], K_PER_CLASS, seed=SEED)
              msgs = build_messages(row["combined"], labels, nc, examples=ex)
          all_messages.append(msgs)

      t0 = time.time()
      preds = await classify_batch(client, con, MODEL, all_messages, labels, SELF_CONSISTENCY_N)
      print(f"Czas: {time.time()-t0:.1f}s")

      metrics = evaluate_and_save(test_df["y"].tolist(), preds, labels, tag, test_df)
      summary_rows.append({
          "experiment": tag, "num_classes": nc, "mode": mode,
          "accuracy": metrics["accuracy"],
          "macro_f1": metrics["macro"]["f1"],
          "weighted_f1": metrics["weighted"]["f1"],
          "unparsed": metrics["n_unparsed"],
      })

  summary = pd.DataFrame(summary_rows)
  print("\n=========== TABELA PORÓWNAWCZA ===========")
  print(summary.to_string(index=False))
  summary.to_csv(os.path.join(OUT_DIR, "summary.csv"), index=False)
  print(f"\nZapisano: {OUT_DIR}/summary.csv + metrics_*.json + cm_*.png + preds_*.csv")

# Colab ma juz dzialajacy loop -> await main()
await main()
```

```python
# =====================================================================
# GPT zero-shot + few-shot na LIAR (2kl i 6kl) - WERSJA Z POPRAWKAMI
# - Structured Outputs (json_schema, enum) -> 0 unparsed
# - K_PER_CLASS=3, SELF_CONSISTENCY_N=3
# Robi 4 eksperymenty: (2kl,zero) (2kl,few) (6kl,zero) (6kl,few)
# =====================================================================

# --------- 0. Instalacja + import ----------
import subprocess, sys
subprocess.run([sys.executable, "-m", "pip", "-q", "install",
                "openai>=1.40.0", "scikit-learn", "pandas", "tqdm", "matplotlib"], check=True)

import os, re, json, hashlib, sqlite3, asyncio, zipfile, urllib.request, getpass, random, time
from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tqdm.auto import tqdm
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.metrics import (precision_recall_fscore_support, classification_report,
                             confusion_matrix, accuracy_score)
from openai import AsyncOpenAI

# --------- 1. KONFIG ----------
MODEL              = "gpt-4o"   # tani; "gpt-4o" mocniejszy ale ~10x drozszy
N_TEST             = 0               # 0 = caly test, inaczej N
K_PER_CLASS        = 3               # 3 przyklady na klase
FEWSHOT_STRATEGY   = "knn"           # "knn" lub "random"
SELF_CONSISTENCY_N = 3               # 3 probki + glosowanie
CONCURRENCY        = 8               # zeby nie walic w 429 RPM
SEED               = 42
OUT_DIR            = "results"
CACHE_DB           = "gpt_cache.sqlite"

EXPERIMENTS = [
    {"num_classes": 2, "mode": "zero-shot"},
    {"num_classes": 2, "mode": "few-shot"},
    {"num_classes": 6, "mode": "zero-shot"},
    {"num_classes": 6, "mode": "few-shot"},
]

# --------- 2. Klucz API ----------
if not os.environ.get("OPENAI_API_KEY"):
    try:
        from google.colab import userdata
        os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
    except Exception:
        os.environ["OPENAI_API_KEY"] = getpass.getpass("OPENAI_API_KEY: ")

random.seed(SEED); np.random.seed(SEED)
os.makedirs(OUT_DIR, exist_ok=True)

# --------- 3. Pobranie LIAR ----------
DATA_DIR = "data"
ZIP_PATH = os.path.join(DATA_DIR, "liar_dataset.zip")
os.makedirs(DATA_DIR, exist_ok=True)
if not os.path.exists(os.path.join(DATA_DIR, "train.tsv")):
    urllib.request.urlretrieve("https://www.cs.ucsb.edu/~william/data/liar_dataset.zip", ZIP_PATH)
    with zipfile.ZipFile(ZIP_PATH) as z:
        z.extractall(DATA_DIR)

COLUMNS = ["id","label","statement","subject","speaker","job_title","state","party",
           "barely_true_counts","false_counts","half_true_counts","mostly_true_counts",
           "pants_on_fire_counts","context"]

def load_split(name):
    df = pd.read_csv(os.path.join(DATA_DIR, f"{name}.tsv"), sep="\t", header=None, names=COLUMNS)
    return df.drop(columns=["id","job_title","state"])

train_raw = load_split("train")
test_raw  = load_split("test")
print(f"train: {train_raw.shape}, test: {test_raw.shape}")

# --------- 4. Combined text (z NaN-safe int) ----------
def combine_text(row):
    speaker = row.get("speaker") if pd.notna(row.get("speaker")) else "unknown"
    party   = row.get("party")   if pd.notna(row.get("party"))   else "unknown"
    def _n(x):
        try:
            v = float(x); return 0 if pd.isna(v) else int(v)
        except (TypeError, ValueError): return 0
    lies   = _n(row.get("false_counts")) + _n(row.get("pants_on_fire_counts"))
    truths = _n(row.get("mostly_true_counts")) + _n(row.get("half_true_counts"))
    parts = [f"{speaker} ({party}, past lies: {lies}, past truths: {truths}): {row['statement']}"]
    if pd.notna(row.get("subject")):
        parts.append("Subject: " + str(row["subject"]))
    if pd.notna(row.get("context")):
        parts.append("Context: " + str(row["context"]))
    return " ".join(parts)

train_raw["combined"] = train_raw.apply(combine_text, axis=1)
test_raw["combined"]  = test_raw.apply(combine_text, axis=1)

# --------- 5. Etykiety ----------
LABELS_6 = ["pants-fire", "false", "barely-true", "half-true", "mostly-true", "true"]
LABEL_DESC_6 = {
    "pants-fire":  "razaco falszywe, absurdalne klamstwo",
    "false":       "falszywe",
    "barely-true": "zawiera ziarno prawdy, ale w istotny sposob mylace",
    "half-true":   "czesciowo prawdziwe, pomija istotne fakty",
    "mostly-true": "w wiekszosci prawdziwe, drobne niescislosci",
    "true":        "prawdziwe",
}
BIN_MAP = {"pants-fire":"fake","false":"fake","barely-true":"fake",
           "half-true":"real","mostly-true":"real","true":"real"}
LABELS_2 = ["fake", "real"]

def prep_split(df, num_classes):
    df = df[df["label"].isin(LABELS_6)].copy()
    if num_classes == 2:
        df["y_str"] = df["label"].map(BIN_MAP); labels = LABELS_2
    else:
        df["y_str"] = df["label"]; labels = LABELS_6
    df["y"] = df["y_str"].map({l:i for i,l in enumerate(labels)})
    return df.reset_index(drop=True), labels

# --------- 6. Prompty ----------
def build_system_prompt(labels, num_classes):
    if num_classes == 2:
        guide = ("- fake: wypowiedz falszywa lub w wiekszosci mylaca\n"
                 "- real: wypowiedz prawdziwa lub w wiekszosci prawdziwa")
    else:
        guide = "\n".join(f"- {l}: {LABEL_DESC_6[l]}" for l in labels)
    return ("Jestes klasyfikatorem prawdziwosci wypowiedzi politycznych ze zbioru LIAR. "
            f"Przypisujesz wypowiedzi DOKLADNIE jedna z {len(labels)} etykiet:\n{guide}\n"
            f"Odpowiadasz w formacie JSON: {{\"label\": \"<jedna z: {', '.join(labels)}>\"}}.")

def fmt_user(combined, labels):
    return f"Wypowiedz:\n{combined}\n\nZwroc JSON z etykieta z: {', '.join(labels)}."

def build_messages(combined, labels, num_classes, examples=None):
    msgs = [{"role":"system","content":build_system_prompt(labels, num_classes)}]
    if examples:
        for ex in examples:
            msgs.append({"role":"user","content":fmt_user(ex["combined"], labels)})
            msgs.append({"role":"assistant","content":json.dumps({"label": ex["y_str"]})})
    msgs.append({"role":"user","content":fmt_user(combined, labels)})
    return msgs

# --------- 7. Few-shot selektor ----------
class FewShotSelector:
    def __init__(self, train_df, labels, strategy="knn"):
        self.train_df = train_df; self.labels = labels; self.strategy = strategy
        if strategy == "knn":
            self.vec = TfidfVectorizer(ngram_range=(1,2), min_df=2, max_features=50000)
            self.train_mat = self.vec.fit_transform(train_df["combined"].tolist())
        self.by_label = {l: train_df[train_df.y_str == l].reset_index(drop=True) for l in labels}
        self.by_label_idx = {l: train_df.index[train_df.y_str == l].to_numpy() for l in labels}

    def select(self, query_combined, k_per_class, seed):
        out = []
        if self.strategy == "random":
            for l in self.labels:
                pool = self.by_label[l]
                if len(pool)==0: continue
                out.extend(pool.sample(min(k_per_class, len(pool)), random_state=seed).to_dict("records"))
        else:
            q = self.vec.transform([query_combined])
            sims = cosine_similarity(q, self.train_mat).ravel()
            for l in self.labels:
                idxs = self.by_label_idx[l]
                if len(idxs)==0: continue
                top = idxs[np.argsort(-sims[idxs])[:k_per_class]]
                out.extend(self.train_df.iloc[top].to_dict("records"))
        random.Random(seed).shuffle(out)
        return out

# --------- 8. STRUCTURED OUTPUTS - schema per zestaw etykiet ----------
LABEL_SCHEMA_CACHE = {}
def labels_to_schema(labels):
    key = tuple(labels)
    if key not in LABEL_SCHEMA_CACHE:
        LABEL_SCHEMA_CACHE[key] = {
            "type": "json_schema",
            "json_schema": {
                "name": "liar_label", "strict": True,
                "schema": {
                    "type": "object",
                    "properties": {"label": {"type": "string", "enum": list(labels)}},
                    "required": ["label"], "additionalProperties": False,
                },
            },
        }
    return LABEL_SCHEMA_CACHE[key]

# --------- 9. Cache SQLite ----------
def cache_init():
    con = sqlite3.connect(CACHE_DB)
    con.execute("CREATE TABLE IF NOT EXISTS c (k TEXT PRIMARY KEY, v TEXT)")
    con.commit(); return con

def cache_key(model, messages, temperature, sample_idx, schema_tag):
    blob = json.dumps([model, messages, temperature, sample_idx, schema_tag], ensure_ascii=False)
    return hashlib.sha1(blob.encode()).hexdigest()

def cache_get(con, k):
    r = con.execute("SELECT v FROM c WHERE k=?", (k,)).fetchone()
    return r[0] if r else None

def cache_put(con, k, v):
    con.execute("INSERT OR REPLACE INTO c VALUES (?,?)", (k, v)); con.commit()

# --------- 10. Parsowanie ----------
def parse_label(text, labels):
    if not text: return None
    try:
        obj = json.loads(text)
        lab = obj.get("label") if isinstance(obj, dict) else None
        if lab in labels: return lab
    except Exception:
        pass
    t = text.strip().lower().strip(".,;:!?\"'`{} ")
    if t in labels: return t
    for l in sorted(labels, key=len, reverse=True):
        if l in t: return l
    return None

# --------- 11. Async runner ze Structured Outputs ----------
async def call_one(client, sem, con, model, messages, temperature, sample_idx, labels, max_retries=5):
    schema_tag = "|".join(labels)
    k = cache_key(model, messages, temperature, sample_idx, schema_tag)
    hit = cache_get(con, k)
    if hit is not None: return hit
    rf = labels_to_schema(labels)
    async with sem:
        for attempt in range(max_retries):
            try:
                resp = await client.chat.completions.create(
                    model=model, messages=messages,
                    response_format=rf,
                    max_tokens=30, temperature=temperature, seed=SEED+sample_idx)
                txt = resp.choices[0].message.content or ""
                cache_put(con, k, txt); return txt
            except Exception as e:
                if attempt == max_retries-1:
                    if "rate_limit" not in str(e).lower():
                        print(f"[API error] {e}")
                    return ""
                await asyncio.sleep(2**attempt + random.random())

def tqdm_wrap(tasks):
    pbar = tqdm(total=len(tasks))
    async def wrap(t):
        r = await t; pbar.update(1); return r
    return [wrap(t) for t in tasks]

async def classify_batch(client, con, model, all_messages, labels, sc_n):
    sem = asyncio.Semaphore(CONCURRENCY)
    if sc_n <= 1:
        tasks = [call_one(client, sem, con, model, m, 0.0, 0, labels) for m in all_messages]
        outs = await asyncio.gather(*tqdm_wrap(tasks))
        return [parse_label(o, labels) for o in outs]
    tasks = []
    for i, m in enumerate(all_messages):
        for s in range(sc_n):
            tasks.append(call_one(client, sem, con, model, m, 0.7, s, labels))
    raw = await asyncio.gather(*tqdm_wrap(tasks))
    preds = []
    for i in range(len(all_messages)):
        votes = [parse_label(raw[i*sc_n + s], labels) for s in range(sc_n)]
        votes = [v for v in votes if v]
        preds.append(Counter(votes).most_common(1)[0][0] if votes else None)
    return preds

# --------- 12. Ewaluacja + zapis ----------
def evaluate_and_save(y_true, y_pred_str, labels, tag, test_df):
    label2id = {l:i for i,l in enumerate(labels)}
    y_pred = [label2id.get(p, -1) for p in y_pred_str]
    mask = [p != -1 for p in y_pred]
    yt = np.array([y for y,m in zip(y_true, mask) if m])
    yp = np.array([y for y,m in zip(y_pred, mask) if m])
    n_unparsed = sum(1 for m in mask if not m)

    idx = list(range(len(labels)))
    per_p, per_r, per_f1, per_sup = precision_recall_fscore_support(
        yt, yp, labels=idx, average=None, zero_division=0)
    macro = precision_recall_fscore_support(yt, yp, labels=idx, average="macro", zero_division=0)
    weighted = precision_recall_fscore_support(yt, yp, labels=idx, average="weighted", zero_division=0)
    acc = accuracy_score(yt, yp)
    cm = confusion_matrix(yt, yp, labels=idx)

    print(f"\n=== {tag} ===  acc={acc:.3f}  macro-F1={macro[2]:.3f}  unparsed={n_unparsed}")
    print(classification_report(yt, yp, labels=idx, target_names=labels, zero_division=0))
    print("Confusion matrix:")
    print(pd.DataFrame(cm, index=labels, columns=labels))

    fig, ax = plt.subplots(figsize=(1.2*len(labels)+2, 1.2*len(labels)+1))
    im = ax.imshow(cm, cmap="Blues")
    ax.set_xticks(range(len(labels))); ax.set_xticklabels(labels, rotation=45, ha="right")
    ax.set_yticks(range(len(labels))); ax.set_yticklabels(labels)
    for i in range(len(labels)):
        for j in range(len(labels)):
            ax.text(j, i, str(cm[i,j]), ha="center", va="center",
                    color="white" if cm[i,j] > cm.max()/2 else "black")
    ax.set_xlabel("Predicted"); ax.set_ylabel("True"); ax.set_title(tag)
    fig.colorbar(im); fig.tight_layout()
    png = os.path.join(OUT_DIR, f"cm_{tag}.png")
    fig.savefig(png, dpi=120); plt.show(); plt.close(fig)

    test_df = test_df.copy()
    test_df["pred"] = y_pred_str
    test_df["correct"] = test_df["y_str"] == test_df["pred"]
    csv_path = os.path.join(OUT_DIR, f"preds_{tag}.csv")
    test_df[["speaker","party","subject","statement","y_str","pred","correct"]].to_csv(csv_path, index=False)

    err = test_df[~test_df["correct"]]
    top_err_subj = err["subject"].value_counts().head(10).to_dict()

    metrics = {
        "tag": tag, "n_total": len(y_true), "n_unparsed": int(n_unparsed),
        "accuracy": float(acc),
        "macro":    {"precision":float(macro[0]),"recall":float(macro[1]),"f1":float(macro[2])},
        "weighted": {"precision":float(weighted[0]),"recall":float(weighted[1]),"f1":float(weighted[2])},
        "per_class": [{"label":l,"precision":float(per_p[i]),"recall":float(per_r[i]),
                       "f1":float(per_f1[i]),"support":int(per_sup[i])} for i,l in enumerate(labels)],
        "confusion_matrix": cm.tolist(), "labels": labels,
        "top_error_subjects": top_err_subj,
    }
    with open(os.path.join(OUT_DIR, f"metrics_{tag}.json"), "w") as f:
        json.dump(metrics, f, ensure_ascii=False, indent=2)
    return metrics

# --------- 13. Glowna petla ----------
async def main():
    client = AsyncOpenAI()
    con = cache_init()
    summary_rows = []

    for exp in EXPERIMENTS:
        nc, mode = exp["num_classes"], exp["mode"]
        tag = f"{MODEL}_{nc}cl_{mode}_{FEWSHOT_STRATEGY if mode=='few-shot' else 'na'}_k{K_PER_CLASS}_sc{SELF_CONSISTENCY_N}_so"
        print(f"\n############ {tag} ############")

        train_df, labels = prep_split(train_raw, nc)
        test_df,  _      = prep_split(test_raw,  nc)
        if N_TEST > 0:
            test_df = test_df.sample(min(N_TEST, len(test_df)), random_state=SEED).reset_index(drop=True)
        print(f"Rozklad testu: {test_df['y_str'].value_counts().to_dict()}")

        selector = FewShotSelector(train_df, labels, FEWSHOT_STRATEGY) if mode == "few-shot" else None
        all_messages = []
        for _, row in test_df.iterrows():
            if mode == "zero-shot":
                msgs = build_messages(row["combined"], labels, nc)
            else:
                ex = selector.select(row["combined"], K_PER_CLASS, seed=SEED)
                msgs = build_messages(row["combined"], labels, nc, examples=ex)
            all_messages.append(msgs)

        t0 = time.time()
        preds = await classify_batch(client, con, MODEL, all_messages, labels, SELF_CONSISTENCY_N)
        print(f"Czas: {time.time()-t0:.1f}s")

        metrics = evaluate_and_save(test_df["y"].tolist(), preds, labels, tag, test_df)
        summary_rows.append({
            "experiment": tag, "num_classes": nc, "mode": mode,
            "accuracy": metrics["accuracy"],
            "macro_f1": metrics["macro"]["f1"],
            "weighted_f1": metrics["weighted"]["f1"],
            "unparsed": metrics["n_unparsed"],
        })

    summary = pd.DataFrame(summary_rows)
    print("\n=========== TABELA PORÓWNAWCZA ===========")
    print(summary.to_string(index=False))
    summary.to_csv(os.path.join(OUT_DIR, "summary.csv"), index=False)
    print(f"\nZapisano: {OUT_DIR}/summary.csv + metrics_*.json + cm_*.png + preds_*.csv")

await main()
```

```python
import subprocess, sys
subprocess.run([sys.executable, "-m", "pip", "-q", "install",
                "openai>=1.40.0", "scikit-learn", "pandas", "tqdm", "matplotlib"], check=True)

import os, re, json, hashlib, sqlite3, asyncio, zipfile, urllib.request, getpass, random, time
from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tqdm.auto import tqdm
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.metrics import (precision_recall_fscore_support, classification_report,
                              confusion_matrix, accuracy_score)
from openai import AsyncOpenAI

# --------- KONFIG ----------
MODEL              = "gpt-4o"
K_PER_CLASS        = 1
FEWSHOT_STRATEGY   = "knn"
SELF_CONSISTENCY_N = 1
CONCURRENCY        = 8
SEED               = 42
OUT_DIR            = "results"
CACHE_DB           = "gpt_cache.sqlite"
random.seed(SEED); np.random.seed(SEED)
os.makedirs(OUT_DIR, exist_ok=True)

# --------- KLUCZ ----------
if not os.environ.get("OPENAI_API_KEY"):
    try:
        from google.colab import userdata
        os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
    except Exception:
        os.environ["OPENAI_API_KEY"] = getpass.getpass("OPENAI_API_KEY: ")

# --------- DANE LIAR ----------
DATA_DIR = "data"
os.makedirs(DATA_DIR, exist_ok=True)
if not os.path.exists(os.path.join(DATA_DIR, "train.tsv")):
    urllib.request.urlretrieve("https://www.cs.ucsb.edu/~william/data/liar_dataset.zip",
                                os.path.join(DATA_DIR, "liar.zip"))
    with zipfile.ZipFile(os.path.join(DATA_DIR, "liar.zip")) as z:
        z.extractall(DATA_DIR)

COLUMNS = ["id","label","statement","subject","speaker","job_title","state","party",
            "barely_true_counts","false_counts","half_true_counts","mostly_true_counts",
            "pants_on_fire_counts","context"]

def load_split(name):
    df = pd.read_csv(os.path.join(DATA_DIR, f"{name}.tsv"), sep="\t", header=None, names=COLUMNS)
    return df.drop(columns=["id","job_title","state"])

train_raw = load_split("train")
test_raw  = load_split("test")
print(f"train: {train_raw.shape}, test: {test_raw.shape}")

def combine_text(row):
    speaker = row.get("speaker") if pd.notna(row.get("speaker")) else "unknown"
    party   = row.get("party")   if pd.notna(row.get("party"))   else "unknown"
    def _n(x):
        try:
            v = float(x); return 0 if pd.isna(v) else int(v)
        except (TypeError, ValueError): return 0
    lies   = _n(row.get("false_counts")) + _n(row.get("pants_on_fire_counts"))
    truths = _n(row.get("mostly_true_counts")) + _n(row.get("half_true_counts"))
    parts = [f"{speaker} ({party}, past lies: {lies}, past truths: {truths}): {row['statement']}"]
    if pd.notna(row.get("subject")): parts.append("Subject: " + str(row["subject"]))
    if pd.notna(row.get("context")): parts.append("Context: " + str(row["context"]))
    return " ".join(parts)

train_raw["combined"] = train_raw.apply(combine_text, axis=1)
test_raw["combined"]  = test_raw.apply(combine_text, axis=1)

LABELS_6 = ["pants-fire", "false", "barely-true", "half-true", "mostly-true", "true"]
LABEL_DESC_6 = {
    "pants-fire":"razaco falszywe","false":"falszywe","barely-true":"mylace",
    "half-true":"czesciowo prawdziwe","mostly-true":"w wiekszosci prawdziwe","true":"prawdziwe",
}

def prep_split(df, num_classes=6):
    df = df[df["label"].isin(LABELS_6)].copy()
    df["y_str"] = df["label"]
    df["y"] = df["y_str"].map({l:i for i,l in enumerate(LABELS_6)})
    return df.reset_index(drop=True), LABELS_6

# --------- PROMPTY ----------
def build_system_prompt(labels):
    guide = "\n".join(f"- {l}: {LABEL_DESC_6[l]}" for l in labels)
    return ("Jestes klasyfikatorem prawdziwosci wypowiedzi politycznych ze zbioru LIAR. "
            f"Przypisujesz wypowiedzi DOKLADNIE jedna z {len(labels)} etykiet:\n{guide}\n"
            f"Odpowiadasz w formacie JSON: {{\"label\": \"<jedna z: {', '.join(labels)}>\"}}.")

def fmt_user(combined, labels):
    return f"Wypowiedz:\n{combined}\n\nZwroc JSON z etykieta z: {', '.join(labels)}."

def build_messages(combined, labels, examples=None):
    msgs = [{"role":"system","content":build_system_prompt(labels)}]
    if examples:
        for ex in examples:
            msgs.append({"role":"user","content":fmt_user(ex["combined"], labels)})
            msgs.append({"role":"assistant","content":json.dumps({"label": ex["y_str"]})})
    msgs.append({"role":"user","content":fmt_user(combined, labels)})
    return msgs

# --------- FEW-SHOT ----------
class FewShotSelector:
    def __init__(self, train_df, labels):
        self.train_df = train_df; self.labels = labels
        self.vec = TfidfVectorizer(ngram_range=(1,2), min_df=2, max_features=50000)
        self.train_mat = self.vec.fit_transform(train_df["combined"].tolist())
        self.by_label_idx = {l: train_df.index[train_df.y_str == l].to_numpy() for l in labels}

    def select(self, query_combined, k_per_class, seed):
        q = self.vec.transform([query_combined])
        sims = cosine_similarity(q, self.train_mat).ravel()
        out = []
        for l in self.labels:
            idxs = self.by_label_idx[l]
            if len(idxs)==0: continue
            top = idxs[np.argsort(-sims[idxs])[:k_per_class]]
            out.extend(self.train_df.iloc[top].to_dict("records"))
        random.Random(seed).shuffle(out)
        return out

# --------- STRUCTURED OUTPUTS ----------
def labels_to_schema(labels):
    return {"type":"json_schema","json_schema":{"name":"liar_label","strict":True,
        "schema":{"type":"object","properties":{"label":{"type":"string","enum":list(labels)}},
                  "required":["label"],"additionalProperties":False}}}

# --------- CACHE ----------
def cache_init():
    con = sqlite3.connect(CACHE_DB)
    con.execute("CREATE TABLE IF NOT EXISTS c (k TEXT PRIMARY KEY, v TEXT)")
    con.commit(); return con

def cache_key(model, messages, temperature, sample_idx, schema_tag):
    blob = json.dumps([model, messages, temperature, sample_idx, schema_tag], ensure_ascii=False)
    return hashlib.sha1(blob.encode()).hexdigest()

def cache_get(con, k):
    r = con.execute("SELECT v FROM c WHERE k=?", (k,)).fetchone()
    return r[0] if r else None

def cache_put(con, k, v):
    con.execute("INSERT OR REPLACE INTO c VALUES (?,?)", (k, v)); con.commit()

# >>> czyszczenie pustych wpisow z poprzedniego (failed) runu <<<
con = cache_init()
n_before = con.execute("SELECT COUNT(*) FROM c").fetchone()[0]
con.execute("DELETE FROM c WHERE v = '' OR v IS NULL")
n_after = con.execute("SELECT COUNT(*) FROM c").fetchone()[0]
con.commit()
print(f"Cache: usunieto {n_before - n_after} pustych wpisow (zostalo {n_after}).")

# --------- PARSOWANIE ----------
def parse_label(text, labels):
    if not text: return None
    try:
        obj = json.loads(text)
        lab = obj.get("label") if isinstance(obj, dict) else None
        if lab in labels: return lab
    except Exception: pass
    t = text.strip().lower().strip(".,;:!?\"'`{} ")
    if t in labels: return t
    for l in sorted(labels, key=len, reverse=True):
        if l in t: return l
    return None

async def call_one(client, sem, con, model, messages, temperature, sample_idx, labels, max_retries=5):
    schema_tag = "|".join(labels)
    k = cache_key(model, messages, temperature, sample_idx, schema_tag)
    hit = cache_get(con, k)
    if hit is not None: return hit
    rf = labels_to_schema(labels)
    async with sem:
        for attempt in range(max_retries):
            try:
                resp = await client.chat.completions.create(
                    model=model, messages=messages, response_format=rf,
                    max_tokens=30, temperature=temperature, seed=SEED+sample_idx)
                txt = resp.choices[0].message.content or ""
                if txt:
                    cache_put(con, k, txt)
                return txt
            except Exception as e:
                if attempt == max_retries-1:
                    print(f"[API error] {e}")
                    return ""
                await asyncio.sleep(2**attempt + random.random())

def tqdm_wrap(tasks):
    pbar = tqdm(total=len(tasks))
    async def wrap(t):
        r = await t; pbar.update(1); return r
    return [wrap(t) for t in tasks]

async def classify_batch(client, con, model, all_messages, labels, sc_n):
    sem = asyncio.Semaphore(CONCURRENCY)
    tasks = []
    for i, m in enumerate(all_messages):
        for s in range(sc_n):
            tasks.append(call_one(client, sem, con, model, m, 0.0 if sc_n == 1 else 0.7, s, labels))
    raw = await asyncio.gather(*tqdm_wrap(tasks))
    preds = []
    for i in range(len(all_messages)):
        votes = [parse_label(raw[i*sc_n + s], labels) for s in range(sc_n)]
        votes = [v for v in votes if v]
        preds.append(Counter(votes).most_common(1)[0][0] if votes else None)
    return preds

# --------- EWALUACJA ----------
def evaluate_and_save(y_true, y_pred_str, labels, tag, test_df):
    label2id = {l:i for i,l in enumerate(labels)}
    y_pred = [label2id.get(p, -1) for p in y_pred_str]
    mask = [p != -1 for p in y_pred]
    yt = np.array([y for y,m in zip(y_true, mask) if m])
    yp = np.array([y for y,m in zip(y_pred, mask) if m])
    n_unparsed = sum(1 for m in mask if not m)
    idx = list(range(len(labels)))
    macro = precision_recall_fscore_support(yt, yp, labels=idx, average="macro", zero_division=0)
    weighted = precision_recall_fscore_support(yt, yp, labels=idx, average="weighted", zero_division=0)
    acc = accuracy_score(yt, yp)
    cm = confusion_matrix(yt, yp, labels=idx)
    print(f"\n=== {tag} ===  acc={acc:.3f}  macro-F1={macro[2]:.3f}  unparsed={n_unparsed}")
    print(classification_report(yt, yp, labels=idx, target_names=labels, zero_division=0))
    print("Confusion matrix:")
    print(pd.DataFrame(cm, index=labels, columns=labels))

    fig, ax = plt.subplots(figsize=(1.2*len(labels)+2, 1.2*len(labels)+1))
    im = ax.imshow(cm, cmap="Blues")
    ax.set_xticks(range(len(labels))); ax.set_xticklabels(labels, rotation=45, ha="right")
    ax.set_yticks(range(len(labels))); ax.set_yticklabels(labels)
    for i in range(len(labels)):
        for j in range(len(labels)):
            ax.text(j, i, str(cm[i,j]), ha="center", va="center",
                    color="white" if cm[i,j] > cm.max()/2 else "black")
    ax.set_xlabel("Predicted"); ax.set_ylabel("True"); ax.set_title(tag)
    fig.colorbar(im); fig.tight_layout()
    fig.savefig(os.path.join(OUT_DIR, f"cm_{tag}.png"), dpi=120); plt.show(); plt.close(fig)
    return {"tag":tag,"accuracy":float(acc),"macro_f1":float(macro[2]),
            "weighted_f1":float(weighted[2]),"unparsed":int(n_unparsed)}

# --------- ODPALENIE: tylko 6cl FS ----------
nc = 6; mode = "few-shot"
tag = f"{MODEL}_{nc}cl_{mode}_{FEWSHOT_STRATEGY}_k{K_PER_CLASS}_sc{SELF_CONSISTENCY_N}_so"
print(f"\n############ {tag} ############")

train_df, labels = prep_split(train_raw)
test_df,  _      = prep_split(test_raw)
print(f"Rozklad testu: {test_df['y_str'].value_counts().to_dict()}")

selector = FewShotSelector(train_df, labels)
all_messages = []
for _, row in test_df.iterrows():
    ex = selector.select(row["combined"], K_PER_CLASS, seed=SEED)
    all_messages.append(build_messages(row["combined"], labels, examples=ex))

async def run():
    client = AsyncOpenAI()
    t0 = time.time()
    preds = await classify_batch(client, con, MODEL, all_messages, labels, SELF_CONSISTENCY_N)
    print(f"Czas: {time.time()-t0:.1f}s")
    return preds

preds = await run()
metrics = evaluate_and_save(test_df["y"].tolist(), preds, labels, tag, test_df)

# --------- DOPISANIE DO SUMMARY ----------
new_row = pd.DataFrame([{**metrics, "num_classes":nc, "mode":mode}])
sum_path = os.path.join(OUT_DIR, "summary.csv")
if os.path.exists(sum_path):
    existing = pd.read_csv(sum_path)
    existing = existing[existing.get("experiment", existing.get("tag")) != tag]
    updated = pd.concat([existing, new_row.rename(columns={"tag":"experiment"})], ignore_index=True)
else:
    updated = new_row.rename(columns={"tag":"experiment"})
updated.to_csv(sum_path, index=False)
print(f"\nZaktualizowano {sum_path}")
print(updated.to_string(index=False))
```
